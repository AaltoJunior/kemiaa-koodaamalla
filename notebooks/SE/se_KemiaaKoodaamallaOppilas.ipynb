### 1. Rita föreningar med SMILES-strängar.

```python
# Med nedanstående kod kan du rita en förening genom att ange dess SMILES-sträng i 
# fältet smiles = ' ' och köra cellen med CTRL + Enter.
# Börja genom att prova en av SMILES-strängarna från tabellen ovan.

# Vi importerar nödvändinga moduler från biblioteken.
from rdkit import Chem
from rdkit.Chem import rdDepictor
from rdkit.Chem.Draw import IPythonConsole
from rdkit.Chem.Draw import rdMolDraw2D
from IPython.display import SVG

smiles = 'C'  # Klistra in önskad SMILES-sträng mellan tecknen ' ' för att ersätta C. Tryck till sist CTRL + ENTER för att köra cellen.
m = Chem.MolFromSmiles(smiles)

# m = Chem.AddHs(m) # Du kan använda detta kommando för att visualisera alla väteatomer om du vill. Ta bort det första #-tecknet 
# för att aktivera inställningen.

# Vi skapar en moltosvg funktion som gör en SVG-bild av strukturformeln från SMILES-strängen.
def moltosvg(mol, molSize=(350, 300), kekulize=True):
    mc = Chem.Mol(mol.ToBinary())
    if kekulize:
        try:
            Chem.Kekulize(mc)
        except:
            mc = Chem.Mol(mol.ToBinary())
    if not mc.GetNumConformers():
        rdDepictor.Compute2DCoords(mc)
    drawer = rdMolDraw2D.MolDraw2DSVG(molSize[0], molSize[1])
    drawer.DrawMolecule(mc)
    drawer.FinishDrawing()
    svg = drawer.GetDrawingText()
    svg = drawer.GetDrawingText()
    return SVG(svg)
    

rakenneKaava = moltosvg(m)  # Definiera variabeln structuralFormula och sätt dess värde till en SVG-bild med funktionen moltosvg.
display(rakenneKaava)       # Visa bilden av strukturformeln.
```

## 2. Skapa en kopplingstabell
Vi testar hur kopplingstabeller fungerar. Välj en mer komplex SMILES-sträng och försök skapa en kopplingstabell för den med hjälp av koden nedan. Du kan också prova med flera SMILES-strängar. Datorn indexerar från noll, det vill säga det första kolet har nummer 0 i tabellen och i strukturformeln.

```python
# Vi importerar nödvändiga moduler från biblioteken.
from rdkit import Chem
from rdkit.Chem import rdDepictor
from rdkit.Chem.Draw import IPythonConsole
from rdkit.Chem.Draw import rdMolDraw2D
from IPython.display import SVG
import numpy as np
import pandas as pd

mol = Chem.MolFromSmiles('C') # Byt ut C mot en mer komplex SMILES-sträng.

# Vi definierar inställningar för rdkits ritfunktion.
d2d = rdMolDraw2D.MolDraw2DSVG(300,300)
d2d.drawOptions().addAtomIndices = True  # Vi tillägger atomindex i bilden. Numreringen börjar på 0 i datorn.
d2d.DrawMolecules([mol])
d2d.FinishDrawing()
svg = SVG(d2d.GetDrawingText())

num_atoms = mol.GetNumAtoms()
connectivity_matrix = np.zeros((num_atoms, num_atoms), dtype="f")
pd.options.display.float_format = lambda x: ('%g' % x)


# Vi skapar en kopplingstabell med hjälp av en for‑loop.
for bond in mol.GetBonds():
    begin_atom_idx = bond.GetBeginAtomIdx()    
    end_atom_idx = bond.GetEndAtomIdx()
    bond_type = (bond.GetBondTypeAsDouble())    
    connectivity_matrix[begin_atom_idx][end_atom_idx] = bond_type
    connectivity_matrix[end_atom_idx][begin_atom_idx] = bond_type
    
df = pd.DataFrame(connectivity_matrix)

# Vi visar strukturformeln och kopplingstabellen.
display(svg, df)
```

```python
# Vi importerar nödvändiga moduler från biblioteken.
import requests

# Vi skapar en funktion som hämtar data från PubChem‑databasen med PUG-REST‑metoden.
# Ändra inte den här delen av koden!
def retrieve_names(compound_names):
    pugrest = "https://pubchem.ncbi.nlm.nih.gov/rest/pug"
    pugoper = "property/CanonicalSMILES"
    pugout = "txt"
    results = {}
    for compound_name in compound_names:
        pugin = "compound/name/" + compound_name
        url = f"{pugrest}/{pugin}/{pugoper}/{pugout}"
        response = requests.get(url)
        if response.status_code == 200:
            results[compound_name] = response.text.strip()  # Vi tar bort onödiga mellanslag.
        else:
            results[compound_name] = f"Error: {response.status_code}"
    return results

# Redigera raden nedan. Lägg till de engelskspråkiga namnen på de läkemedelssubstanser du valt inom citattecken.
compound_names = ["", "", ""] 
# Redigera endast raden ovan.

# Hämta SMILES-strängar med funktionen retrieve_names.
canonical_smiles = retrieve_names(compound_names)

# Vi skriver ut resultaten.
for compound_name, smiles in canonical_smiles.items():
    print(f"Canonical SMILES for {compound_name}: {smiles}")
```

## 6. Bekanta dej med IUPAC-namn och PubChem-databasen

```python
# Med den här koden kan du hämta specifik information från PubChem‑databasen med olika trivial- eller IUPAC‑namn.

# Vi importerar nödvändiga moduler från biblioteken.
import pubchempy as pcp 
import pandas as pd
pd.options.display.max_colwidth = 1000 # Vi anger ett längre maxvärde för kolumnbredd i pandas-biblioteket.

# Lista med exempelämnen. Du kan lägga till valfria föreningar här.
names = ["water", 
         "benzene", 
         "methanol", 
         "ethene", 
         "ethanol", 
         "propene", 
         "1-propanol", 
         "2-propanol", 
         "butadiene", 
         "1-butanol", 
         "2-butanol", 
         "tert-butanol"]

# Använd en for‑loop för att gå igenom listan och spara ämnenas ID:n i en lista.
# Föreningar i PubChem‑databasen är indexerade (ID), och CID är förkortningen för kemiskt index.
# De är unika för posterna och fungerar som en tydligare grund för sökning än SMILES eller annan data.
compound_ids = []
for name in names:
    try:
        cp = pcp.get_compounds(name, namespace='name')
        if cp:
            compound_ids.append(cp[0].cid)
    except Exception as e:
        print(f"Error retrieving CID for '{name}', : {e}")
results = pcp.get_properties(['IUPACName', 'MolecularFormula', 'MolecularWeight', 'CanonicalSMILES'], compound_ids)

# Vi skapar en tabell med hjälp av pandas-biblioteket och skriver ut den.
df = pd.DataFrame(results)
display(df)
```

### 8. Anpassa en linjär regressionsmodell

Linjär anpassning är troligen bekant från mattelektionerna. Denna typ av matematisk modellering är en av de centrala metoderna inom kemoinformatik, eftersom den låter oss undersöka sambandet (korrelationen) mellan olika variabler, från två och uppåt. Nedan finns kod som kan användas för att anpassa en regressionsmodell till en dataset. Vad lägger du märke till när koden skriver ut en graf?

```python
# Vi importerar nödvändiga moduler från biblioteken.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression

# Vi hämtar data från .csv-fil som en tabell
df = pd.read_csv('../../data/BP.csv')

# Vi anpassar en linjär regressionsmodell
model = LinearRegression()
X = df[['Moolimassa']]          # Oberoende variabel
y = df[['Kiehumispiste_K']]     # Beroende variabel
model.fit(X, y)

# Vi ritar ett punktdiagram av data
plt.figure()
sns.scatterplot(x='Moolimassa', y='Kiehumispiste_K', data=df, color='tab:blue', label='Datapunkter')

# Vi ritar regressionslinjen
plt.plot(X, model.predict(X), color='tab:red', label='Linjär regression')

# Vi anger axelrubriker och titel
plt.xlabel('Molmassa')
plt.ylabel('Kokpunkt (K)')
plt.title('Linjär regressionsmodell')

# Visa teckenförklaringen
plt.legend()

# Visa diagrammet
plt.show()
```
